```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

```python
df = pd.read_csv('sales_data.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
print('Rows and columns:', df.shape)
```

```python
print('\nColumn names : ')
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print('Duplicate rows:', df.duplicated().sum())
```

```python
df[df['Payment_Method'].isnull()].head()
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
print('Missing values after handling:')
print(df.isnull().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
print('Data type:', df['Date'].dtype)
print(df['Date'].head())
```

```python
#Date range check krna
print('Date range:', df['Date'].min(), 'to', df['Date'].max())
```

```python
#Month column banana
df['Month'] = df['Date'].dt.to_period('M').astype(str)
# .dt dates par operation karne ke liye use hota hai.
#to_period('M') date ko year-month period mein convert krta h.
# ye period values ko string yani text mein convert karta h.
```

```python
df[['Date', 'Month']].head()
```

```python
print('Number of transactions:', len(df))
```

```python
#Total units sold
print('Total units sold:', df['Units_Sold'].sum())
```

```python
#Sales before discount
print('Sales before discount: ₹{:,.2f}'.format(df['Total_Amount'].sum())) 
```

```python
#Sales after discount
print('Sales after discount: ₹{:,.2f}'.format(df['Final_Amount'].sum())) 
```

```python
#Difference: Total_Amount discount se pehle ki amount h, jbki Final_Amount discount ke baad ki amount h
```

```python
#Average final amount pr transaction
print('Average final amount per transaction: ₹{:,.2f}'.format(df['Final_Amount'].mean()))
```

```python
#isse pata chlta h ki ek transaction me average kitni sales amount generate hui.
```

```python
#Average customer rating
print('Average customer rating:', round(df['Customer_Rating'].mean(), 2))
```

```python
# .mean() all customer rating ka average niklata h.
# round(..., 2) result ko decimal ke baad 2 digits tak round karta h.
# Agar rating 4, 5, 3, 4 hain, toh average rating 4.0 hogi.
```

```python
department_sales = df.groupby('Department')['Final_Amount'].sum().sort_values(ascending = False)

display(department_sales)
department_sales.plot(kind='bar', color = 'steelblue') #parameter

plt.title('Sales After Discount by Departemnt')
plt.xlabel('Department')
plt.ylabel('Final_Amount (₹)')
plt.xticks(rotation=35, ha='right')
plt.tight_layout() # Graph ka layout adjust krna
plt.show()
```

```python
# Transaction by region

region_counts = df['Region'].value_counts()
display(region_counts)

region_counts.plot(kind='bar', color = 'darkseagreen')
plt.title('Number of Transactions by Region')
plt.xlabel('Region')
plt.ylabel('Transaction Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Payment method analysis (transaction ki count dikhata)

payment_counts = df['Payment_Method'].value_counts()
display(payment_counts)

payment_counts.plot(kind='bar', color='cornflowerblue')
plt.title('Transactions by Payment Method')
plt.xlabel('Payment_Method')
plt.ylabel('Transaction Count')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
#Discount analysis
#Dekho ki har alag discount % kitni baar use hua h.

discount_counts = df['Discount_Percent'].value_counts().sort_index()
display(discount_counts)

sns.countplot(data=df, x ='Discount_Percent')

plt.title('Transactions by Discount Percentage')
plt.xlabel('Discount (%)')
plt.ylabel('Transaction Count')
plt.tight_layout()
plt.show()
```

```python
#Customer ratings
ratng_counts = df['Customer_Rating'].value_counts().sort_index()
display(ratng_counts)

sns.countplot(data = df, x ='Customer_Rating')
plt.title('Customer Rating Disribution')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Transaction')
plt.tight_layout() # for spacing
plt.show()
```

```python
# Highest sales department

top_department = department_sales.idxmax()
print(top_department)

print(f'Highest sales department: {top_department} '
      f'(₹{department_sales.max():,.2f})')
```

```python
top_region_by_transactions = region_counts.idxmax()

print(f'Most transactions by region: {top_region_by_transactions} '
      f'({region_counts.max()} transactions)')
```

```python
top_payment_method = payment_counts.idxmax()
print(f'Most common payment method: {top_payment_method} '
      f'({payment_counts.max()} transactions)')
print(f'Average customer rating: {df['Customer_Rating'].mean():.2f} out of 5')
```

```python
# Monthly sales trend
monthly_sales = df.groupby('Month')['Final_Amount'].sum()
display(monthly_sales)

# Figure size set karein
plt.figure(figsize=(8, 5))

# Bar chart plot karein with custom colors
monthly_sales.plot(
    kind='bar', color=['#4C72B0', '#55A868'], width=0.4, 
    edgecolor='black'
)

# Title aur Labels
plt.title('Monthly Sales Comparison (Jan vs Feb)', fontsize=14,
          pad=15)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Final Amount (₹)', fontsize=12)

# X-axis ke labels ko seedha rakhne ke liye
plt.xticks(rotation=0)

# Grid lines on y-axis for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Layout adjust karein aur show karein
plt.tight_layout()
plt.show()
```

```python
#1. Find the department with the highest **number of units sold** (not sales amount).
#(department-wise total Units_Sold)

department_units = df.groupby('Department')['Units_Sold'].sum()

print(department_units)

print('Highest Units Sold Department:', department_units.idxmax())
print('Total Units Sold:',department_units.max())

department_units.plot(kind = 'bar', color='steelblue')

plt.title('Department-wise Total Units Sold')
plt.ylabel('Total Units Sold')
plt.xlabel('Department')
plt.xticks(rotation=45)
plt.show()
```

```python
payment_sales = df.groupby('Payment_Method')['Final_Amount'].sum()

print(payment_sales)
print('Highest Sales Payment Method :', payment_sales.idxmax())
print('Total Final Sales:', payment_sales.max())

payment_sales.plot(kind = 'bar', color='steelblue')
plt.title('Total Final Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Final Sales (₹)')
plt.xticks(rotation=0)
plt.show()
```

```python
# Find the Region with the highest total final_amount.

region_sales = df.groupby('Region')['Final_Amount'].sum()

print(region_sales)

print('Highest Sales Region:', region_sales.idxmax())
print('Total Final Sales:', region_sales.max())

region_sales.plot(kind='bar', color='steelblue')

plt.title('Region-wise Total Final Sales')
plt.xlabel('Region')
plt.ylabel('Total Final Sales (₹)')
plt.xticks(rotation=0)
plt.show()
```

```python
department_rating = df.groupby('Department')['Customer_Rating'].mean()

print(department_rating)
print('Highest Rated Department:', department_rating.idxmax())
print('Highest Average Rating:', department_rating.max())

department_rating.plot(kind='bar', color='steelblue')

plt.title('Average Customer Rating by Department')
plt.xlabel('Deaprtment')
plt.ylabel('Average Customer Rating')
plt.xticks(rotation=45)
plt.ylim(0,5)
plt.show()
```
